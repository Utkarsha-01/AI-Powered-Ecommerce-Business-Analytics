# ==========================================
# ORDERS - DATE AND AMOUNT VALIDATION
# ==========================================

```python
from numpy._core import numeric
# 1. Convert order_date to datetime
orders["order_date"] = pd.to_datetime(
    orders["order_date"],
    errors="coerce"
)

# 2. Check invalid dates
print("Invalid order dates: ",
      orders["order_date"].isna().sum())

# 3. Check order date range
print("\nOrder date range:")
print(orders["order_date"].min())
print(orders["order_date"].max())

# 4. Check numeric columns
numeric_columns = [
    "discount_percentage",
    "subtotal",
    "shipping_fee",
    "tax_amount",
    "final_amount"
]

print("\nNumeric column summary:")
display(orders[numeric_columns].describe())

# 5. Check negative values
print("\nNegative Values:")
for column in numeric_columns:
  print(column, ":", (orders[column] < 0).sum())

# 6. Check discount percentage range
print("\nDiscount percentage range: ")
print(orders["discount_percentage"].min())
print(orders["discount_percentage"].max())
```

# ==========================================
# ORDERS - FINANCIAL CONSISTENCY CHECK
# ==========================================

```python
# Calculate discount amount
orders["discount_amount"] = (
    orders["subtotal"] *
    orders["discount_percentage"] / 100
)

# Candidate 1: Subtotal _ shipping + tax
orders["calculated_amount_1"] = (
    orders["subtotal"] +
    orders["shipping_fee"] +
    orders["tax_amount"]
)

# Candidate 2: Discounted Subtotal + shipping + tax
orders["calculated_amount_2"] = (
    orders["subtotal"] -
    orders["discount_amount"] +
    orders["shipping_fee"] +
    orders["tax_amount"]
)

# Compare calculated amounts with final_amount
orders["difference_1"] = (
    orders["final_amount"] -
    orders["calculated_amount_1"]
)

orders["difference_2"] = (
    orders["final_amount"] -
    orders["calculated_amount_2"]
)

print("Candidate 1 - Amount difference: ")
display(orders["difference_1"].describe())

print("\nCandidate 2 - Amount difference: ")
display(orders["calculated_amount_2"].describe())

# How many recored match each formula within 1 rupee?
print(
    "\nCandidate 1 matches: ",
    (orders["difference_1"].abs() < 1).sum()
)

print(
    "Candidate 2 matches:",
    (orders["difference_2"].abs() < 1).sum()
)
```

# ==========================================
# MONTHLY E-COMMERCE BUSINESS ANALYSIS
# ==========================================

```python
# Create a month column
orders["order_month"] = (
    orders["order_date"].dt.to_period("M")
)

# 1. Overall order status analysis
status_analysis = (
    orders.groupby("order_status")
    .agg(
        total_orders = ("order_id", "count"),
        total_order_amount = ("final_amount", "sum"),
        average_order_amount = ("final_amount", "mean")
    )
    .sort_values("total_orders", ascending = False)
)

print("Order Status Analysis:")
display(status_analysis.round(2))

# 2. Monthly order trends - all orders
monthly_orders = (
    orders.groupby("order_month")
    .agg(
        total_orders = ("order_id", "count"),
        total_order_amount = ("final_amount", "sum")
    )
    .reset_index()
)

print("\nMothly Order Trends:")
print(monthly_orders.head(12))

# 3. Monthly delivered order performance
delivered_orders = orders[
    orders["order_status"] == "Delivered"
].copy()

monthly_delivered = (
    delivered_orders.groupby("order_month")
    .agg(
        delivered_orders = ("order_id", "count"),
        delivered_order_amount = ("final_amount", "sum")
    )
    .reset_index()
)

print("\nMonthly Delivered Performance: ")
display(monthly_delivered.head(12))
```

```python
import matplotlib.pyplot as plt

# Sort months chronologically
monthly_delivered = monthly_delivered.sort_values(
    "order_month"
).reset_index(drop=True)

# Convert month to text for plotting
monthly_delivered["month_label"] = (
    monthly_delivered["order_month"].astype(str)
)

# ==========================================
# CHART 1: MONTHLY DELIVERED ORDER VALUE
# ==========================================

plt.figure(figsize=(15,6))

plt.plot(
    monthly_delivered["month_label"],
    monthly_delivered["delivered_order_amount"],
    marker = "o"
)

plt.title("Monthly Delivered Order Amount")
plt.xlabel("Month")
plt.ylabel("Delivered Order Amount")

plt.xticks(rotation = 90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ==========================================
# CHART 2: MONTHLY DELIVERED ORDERS
# ==========================================

plt.figure(figsize=(15,6))

plt.plot(
    monthly_delivered["month_label"],
    monthly_delivered["delivered_orders"],
    marker = "o"
)

plt.title("Monthly Delivered Orders")
plt.xlabel("Month")
plt.ylabel("Number of Delivered Orders")

plt.xticks(rotation = 90)
plt.grid(True, alpha = 0.3)
plt.tight_layout()
plt.show()
```

```python
# Extract year from the month
monthly_delivered["year"] = (
    monthly_delivered["order_month"].dt.year
)

# Annual delivered performance
annual_performance = (
    monthly_delivered.groupby("year")
    .agg(
        delivered_orders =("delivered_orders", "sum"),
        delivered_order_amount =(
            "delivered_order_amount", "sum"
        )
    )
    .reset_index()
)

# average order amount
annual_performance["average_order_amount"] = (
    annual_performance["delivered_order_amount"]/
    annual_performance["delivered_orders"]
)

print("Annual Delivered Performance: ")

display(annual_performance.round(2))
```

```python
# Product-level performance
product_performance = (
    product_analysis.groupby(
        ["product_id", "product_name", "brand"]
    )
    .agg(
        units_sold=("quantity", "sum"),
        total_revenue=("item_revenue", "sum"),
        total_profit=("profit", "sum"),
        order_count=("order_id", "nunique")
    )
    .reset_index()
)

# Sort by revenue
top_products = product_performance.sort_values(
    "total_revenue", ascending=False
)

print("Top 10 Products by Revenue:")
display(top_products.head(10).round(2))
```

```python
# Analyze performance by product category
category_performance = (
    product_analysis.groupby("product_type")
    .agg(
        units_sold = ("quantity", "sum"),
        total_revenue = ("item_revenue", "sum"),
        total_profit = ("profit", "sum"),
        product_count = ("product_id", "nunique"),
        order_count = ("order_id", "nunique")
    )
    .reset_index()
)

# Sort categories by revenue
category_performance = category_performance.sort_values(
    "total_revenue", ascending=False
)

print("Product Category Performance")
display(category_performance.round(2))
```

```python
# Calculate profit margin percentage
category_performance["profit_margin"] = (
    category_performance["total_profit"]
    / category_performance["total_revenue"]
    * 100
)

# Sort by profit margin
category_performance = category_performance.sort_values(
    "profit_margin", ascending=False
)

print("Category Profit Margin Analysis")
display(
    category_performance[
        [
            "product_type",
            "total_revenue",
            "total_profit",
            "profit_margin"
        ]
    ].round(2)
)
```

```python
import matplotlib.pyplot as plt

# Chart 1: Revenue by category
plt.figure(figsize=(8, 5))

plt.bar(
    category_performance["product_type"],
    category_performance["total_revenue"],
    color = "lightpink"
)

plt.title("Delivered Item Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenuue")
plt.xticks(rotation = 30)
plt.tight_layout()
plt.show()
```

```python
# Group campaign metrics by marketing channel
channel_analysis = (
    campaigns.groupby("channel")
    .agg(
        campaign_count = ("campaign_id", "count"),
        total_conversions = ("conversions", "sum"),
        total_revenue = ("revenue_generated", "sum"),
        total_cost = ("campaign_cost" ,"sum"),
        total_impressions = ("impressions", "sum"),
        total_clicks = ("clicks", "sum"),
        average_roi = ("roi", "mean")
    )
    .reset_index()
)

# calculate funnel metrics
channel_analysis["click_through_rate"] = (
    channel_analysis["total_clicks"]
    / channel_analysis["total_impressions"]
) * 100

channel_analysis["conversion_per_click_percentage"] = (
    channel_analysis["total_conversions"]
    / channel_analysis["total_clicks"]
) * 100

# Sort by total revenue
channel_analysis = channel_analysis.sort_values(
    "total_revenue",
    ascending = False
)

display(channel_analysis.round(2))
```

```python
# ==========================================
# CUSTOMER ANALYTICS - ORDER FREQUENCY
# ==========================================

# 1. Select delivered orders
delivered_orders = orders[
    orders["order_status"] == "Delivered"
].copy()

print("Total Delivered Orders: ", len(delivered_orders))

# 2. Count delivered orders for each customer
customer_order_frequency = delivered_orders.groupby(
    "customer_id"
).size().reset_index(name = "delivered_orders")

print("\nCustomer Order Frequency")
display(customer_order_frequency.head())

# 3. Categorize customers based on order frequency
def customer_category(order_count):

  if order_count == 1:
    return "One-Tome Customer"

  elif order_count <= 3:
    return "Occasional Customer"

  elif order_count <= 5:
    return "Regular Customer"

  else:
    return "Frequent Customer"

customer_order_frequency["customer_category"] = (
    customer_order_frequency["delivered_orders"]
    .apply(customer_category)
)

# 4. Count customers in each category
customer_category_analysis = (
    customer_order_frequency["customer_category"]
    .value_counts()
    .reset_index()
 )

customer_category_analysis.columns = [
    "Customer Category",
    "Number of Customers"
 ]

 # Calculate Percentages
customer_category_analysis["Percentage"] = (
  customer_category_analysis["Number of Customers"]
/ customer_category_analysis["Number of Customers"].sum()
* 100
 )

# 5. Display results
print("\nCUSTOMER CATEGORY ANALYSIS")
display(customer_category_analysis)

# 6. Calculate repeat customer metrics
total_customers_with_delivered_orders = (
    customer_order_frequency["customer_id"].nunique()
)

repeat_customers = customer_order_frequency[
    customer_order_frequency["delivered_orders"] > 1
]

total_repeat_customers = len(repeat_customers)

repeat_customer_percentage = (
    total_repeat_customers
    / total_customers_with_delivered_orders
    * 100
)

print("\nREPEAT CUSTOMER SUMMARY")
print("-" * 40)

print(
    "Customer with Delivered Orders:",
    total_customers_with_delivered_orders
)

print(
    "Repeat Customers:",
    total_repeat_customers
)

print(
    "Repeat Customer Percentage:",
    round(repeat_customer_percentage,2),
    "%"
)
```
